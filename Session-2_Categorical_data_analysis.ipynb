```python
# read dataset
path1= r'Visadataset.csv'
visa_df = pd.read_csv(path1)
visa_df

# divide into cat & num
cat = visa_df.select_dtypes(include = 'object').columns
num = visa_df.select_dtypes(exclude = 'object').columns

# count of continents
labels = visa_df['continent'].unique().tolist()
count = []
for i in labels:
    con = visa_df['continent'] == i
    count.append(len(visa_df[con]))
# create the dataframe
continent_df = pd.DataFrame(zip(labels,count),columns = ['continent','No of Applicants'])

# save the dataset
continent_df.to_csv('continent_df.csv',index=False)
```
